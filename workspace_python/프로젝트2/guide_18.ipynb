```python
query = """
    SELECT
        *,
        strptime(
            printf('%s %s', 
                strftime(Date::TIMESTAMP, '%Y-%m-%d'),
                replace(replace(Time, '오전', 'AM'), '오후', 'PM')
            ),
            '%Y-%m-%d %p %I:%M:%S.%g'
        ) AS Datetime
    FROM tbl
"""

con = duckdb.connect()
tbl2 = con.execute(query).df()
print(tbl2[['Date', 'Time', 'Datetime']].head())
```
