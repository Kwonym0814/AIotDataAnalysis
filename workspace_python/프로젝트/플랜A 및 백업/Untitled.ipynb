
# [Slide] 지능형 비용 최적화 결과

**Subtitle: Optuna 기반 가동률 정밀 제어 알고리즘**

### 1. 핵심 지표 (Center Focus)

* 
**$1^{st}$ Priority**: **최적 가동률 도출** 


* 
**Status**: **흑자 시나리오 확보** (전기료 + 인건비 + 페널티 통합 최적화) 



### 2. 최적화 전략 (Optimization Strategy - Left)

* 
**지능형 탐색**: 수만 가지의 조업 조합 중 가장 비용이 낮은 '황금 지점'을 AI가 자동 탐색 


* 
**하이브리드 엔진**: 빠른 전체 탐색(Grid) 후 Optuna를 통한 소수점 단위의 정밀 타격 


* 
**수익 모델 통합**: 복잡한 한전 요금제와 역률 규정을 하나의 수익 공식으로 통합 관리 



### 3. 최적화 로직 흐름 (Logic Flow - Right)

1. **단가 분석 (Price Analysis)**: 시간대별 변동하는 전기 요금 확인
2. **상태 파악 (Status Check)**: 실시간 조업량 및 누적 역률 데이터 분석 


3. 
**정밀 제어 (Optuna Control)**: 비용을 최소화하는 최적의 가동률 결정 



### 4. 기대 효과 (System Benefit - Bottom)

* **비용 절감 (Cost Saving)**: 피크 시간대 조업 효율화를 통한 기본 요금 및 전력량 요금 동시 절감
* 
**규제 대응 (Compliance)**: 한전 역률 페널티 구간을 완벽히 회피하여 추가 지출 방어
